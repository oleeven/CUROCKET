# Comparing ROCKET with CUROCKET and varaiates

```python
import numpy as np
from time import time

from sktime.transformations.panel.rocket import Rocket
from CuRocketUnivariate.CuRocket import CuRocketUnivariate as CuRocketUni
from CuRocketMultivariate.CuRocket import CuRocketMultivariate as CuRocketMulti
from CuRocketMultivariateMultiGPU.CuRocket import CuRocketMultivariateMultiGPU as CuRocketMVMG
```

## ROCKET vs CUROCKET (univariate)

```python
n_instances = 1000
n_channels = 1
n_timepoints = 50000
n_kernels = 10000
X = np.random.random((n_instances, n_channels, n_timepoints))

rocket = Rocket(1000, n_jobs = -1)
rocket.fit(X)
t = time()
X_t = rocket.transform(X)
t_rocket= time() - t 
print(f"Time for ROCKET (CPU): {t_rocket:.2f}s")

curocket = CuRocketUni(10000, random_state=1)
curocket.fit(X)
t = time()
X_t_2 = curocket.transform(X)
t_curocket =  time() - t 
print(f"Time for CUROCKET (GPU): {t_curocket:.2f}s.")
print(f"CUROCKET is {t_rocket / t_curocket} times faster than ROCKET")
```
